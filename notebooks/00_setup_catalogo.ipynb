```python
# -------------------------------------------------------
# Setup inicial del proyecto - se corre una sola vez
# Crea el catálogo y los tres schemas de la arquitectura
# -------------------------------------------------------

print("Preparando el espacio de trabajo en Unity Catalog...")
print("=" * 50)

# el catálogo raíz del proyecto
spark.sql("CREATE CATALOG IF NOT EXISTS nyc_taxi_andres")
print("Catálogo principal creado: nyc_taxi_andres")

# capa raw: los datos tal como llegan, sin tocar nada
spark.sql("CREATE SCHEMA IF NOT EXISTS nyc_taxi_andres.raw")
print("Schema 'raw' listo — acá llegan los datos crudos")

# capa trusted: ya limpios y validados
spark.sql("CREATE SCHEMA IF NOT EXISTS nyc_taxi_andres.trusted")
print("Schema 'trusted' listo — datos limpios")

# capa refined: los indicadores finales para el negocio
spark.sql("CREATE SCHEMA IF NOT EXISTS nyc_taxi_andres.refined")
print("Schema 'refined' listo — acá van los KPIs")

# creamos un volumen en Unity Catalog para guardar los archivos fuente
spark.sql("CREATE VOLUME IF NOT EXISTS nyc_taxi_andres.raw.archivos_fuente")
print("Volumen creado en nyc_taxi_andres.raw.archivos_fuente")

print("=" * 50)
print("Ya podemos trabajar con el pipeline")
```
